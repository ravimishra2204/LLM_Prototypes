```python
import transformers
from  transformers import pipeline

# Initialize the zero-shot classification pipeline with a suitable model
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

# Define the text to classify
text_to_classify = "I want to book a flight to London for next week if its not rainy else will go for a musical concert."

# Define candidate labels for intent classification
candidate_labels = ["booking flight", "checking weather", "setting alarm", "playing music", "ordering food"]

# Perform the classification
output = classifier(text_to_classify, candidate_labels)

# Print only the top predicted intent
print(f"The predicted intent is: {output['labels'][0]}")
print(output)
```
